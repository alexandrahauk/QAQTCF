```python
#
# Greedy algorithm to generate maximal bipartite graph with fixed # of vertices Q and girth >= 2N+2
#

import igraph as ig
import numpy as np

Q = 24 # number of qubits we have to work with--this is the number of vertices on the graph
Qh = int(Q/2)
N = 2 # particle number--we will have to generate a graph with girth at least >= 2N+2
min_girth = 2*N + 2

trials = 10**3 # amount of times we should try to generate a graph

types = ([0] * Qh) + ([1] * Qh)

best_graph = ig.Graph.Bipartite(types, [])

for i in range(trials):
    graph = ig.Graph.Bipartite(types, [])
    edge_order = [(i, j) for i in range(Qh) for j in range(Qh, Q)]
    np.random.shuffle(edge_order)

    for edge_index in range(len(edge_order)):
        # add next edge in order
        graph.add_edge(edge_order[edge_index][0], edge_order[edge_index][1])

        # if it makes the girth smaller than the threshold, remove it
        if(graph.girth() < min_girth):
            graph.delete_edges(graph.ecount() - 1)

    if (graph.ecount() > best_graph.ecount()):
        best_graph = graph

print(best_graph.ecount())
```
